```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
file_path = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
attributes = ['sepal-length', 'sepal-width', 'petal-length', 'petal-width', 'class']
df = pd.read_csv(file_path, names=attributes)
```

## Visualisasi menggunakan Pandas

```python
df.plot()
```

```python
df.plot.hist(bins=10)
```

```python
df.groupby("class").mean().plot(kind="bar")
```

```python
df.plot(kind="scatter", x="sepal-length", y="sepal-width")
```

## Visualisasi menggunakan Matplotlib

```python
indeks = df.index
data = df.values

plt.plot(indeks, data[:, :-1])
```

```python
data = df.values[:, :-1]
bins = np.linspace(np.min(data.ravel()), np.max(data.ravel()), 10)

plt.hist(data[:, 0], bins, label="sepal-length")
plt.hist(data[:, 1], bins, label="sepal-width")
plt.hist(data[:, 2], bins, label="petal-length")
plt.hist(data[:, 3], bins, label="petal-width")
plt.show()
```

```python
data = df.values

plt.scatter(data[:, 0], data[:, 1])
```
